#### 3. What are the top 10 average MSRP (Price) per car brand (rounded figure)? with Horizontal Bar Plot

```python
msrp_ee = df.groupby('Make')['MSRP'].mean().head(10)
plt.figure(figsize=(10,6))
msrp_ee.sort_values().plot(kind='barh', color='skyblue')
plt.xlabel('Average MSRP ($)')
plt.title('Top 10 Average MSRP by Car Brand')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

#### 4. How has the MSRP (price) changed over the years (rounded figure)? with Line Plot

```python
import pandas as pd
import matplotlib.pyplot as plt

# Group by Year and calculate average MSRP (rounded)
msrp_by_year = df.groupby('Year')['MSRP'].mean().round()

# Plotting
plt.figure(figsize=(10,6))
plt.plot(msrp_by_year.index, msrp_by_year.values, marker='o', linestyle='-', color='green')
plt.xlabel('Year')
plt.ylabel('Average MSRP ($)')
plt.title('Change in Average MSRP Over the Years')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

#### 5. How Engine HP (power) and MPG (efficiency) have changed over the years? show with two line plots

Hint: Average of (city mpg + highway MPG)

```python

# Step 1: Create Average MPG column
df['Average MPG'] = (df['city mpg'] + df['highway MPG']) / 2

# Step 2: Group by Year and calculate average Engine HP and Average MPG
avg_stats_by_year = df.groupby('Year')[['Engine HP', 'Average MPG']].mean().round()

# Step 3: Plotting
plt.figure(figsize=(12,6))

# Line plot for Engine HP
plt.plot(avg_stats_by_year.index, avg_stats_by_year['Engine HP'], marker='o', label='Average Engine HP', color='blue')

# Line plot for Average MPG
plt.plot(avg_stats_by_year.index, avg_stats_by_year['Average MPG'], marker='o', label='Average MPG (Efficiency)', color='orange')

# Chart settings
plt.title('Change in Engine HP and MPG Over the Years')
plt.xlabel('Year')
plt.ylabel('Value')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

#### 6. What percentage of cars are Compact, Midsize, or Large? Show with Pie chart

```python

# Filter for Compact, Midsize, and Large cars
filtered = df[df['Vehicle Size'].isin(['Compact', 'Midsize', 'Large'])]

# Count each type and calculate percentage
size_counts = filtered['Vehicle Size'].value_counts()
size_percent = (size_counts / len(df)) * 100  # divide by total number of cars

# Plotting
plt.figure(figsize=(4,4))
plt.pie(size_percent, labels=size_percent.index, autopct='%1.1f%%', colors=['#1f77b4', '#ff7f0e', '#2ca02c'], startangle=140)
plt.title('Percentage of Compact, Midsize, and Large Cars')
plt.axis('equal')  # keeps the pie circular
plt.show()
```

#### 7. Which vehicle styles are sold the most? show with vertical bar plot

```python

# Count of each vehicle style
style_counts = df['Vehicle Style'].value_counts()

# Plotting
plt.figure(figsize=(12,6))
style_counts.plot(kind='bar', color='skyblue')
plt.title('Most Commonly Sold Vehicle Styles')
plt.xlabel('Vehicle Style')
plt.ylabel('Number of Vehicles')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

#### 8. Which car brands have the most vehicles priced at $40,000 or more and with a popularity score of at least 1000?

```python

# Step 1: Filter based on price and popularity
filtered_df = df[(df['MSRP'] >= 40000) & (df['Popularity'] >= 1000)]

# Step 2: Count vehicles per brand
brand_counts = filtered_df['Make'].value_counts()

# Step 3: Show the result
print("Brands with most vehicles priced ≥ $40,000 and popularity ≥ 1000:")
print(brand_counts)
```
