```python
# function
def sql_query(q):
    return pd.read_sql_query(q,conn)
```

```python
q = '''
select * from Salaries'''
sql_query(q)
```

##### 1. Top 10 highest paid employees (by TotalPayBenefits)

```python
q = '''select EmployeeName, JobTitle, TotalPayBenefits
      from Salaries order by 3 DESC limit 10'''

sql_query(q)
```

#### 2. Find which year has missing data of Employees with 'Not Provided'/ 'Not provided' .

```python
q = '''SELECT Year, COUNT(*) AS Missing_Count
FROM Salaries
WHERE EmployeeName = 'Not Provided' OR EmployeeName = 'Not provided'
GROUP BY Year
ORDER BY Year'''
sql_query(q)
```

```python
###3. Number of unique agencies

q = ''' select distinct agency from Salaries'''
sql_query(q)
```

```python
### 4 .Show Average BasePay, OvertimePay, and TotalPay for year 2011

q = ''' SELECT 
    AVG(BasePay) AS Avg_BasePay,
    AVG(OvertimePay) AS Avg_OvertimePay,
    AVG(TotalPay) AS Avg_TotalPay
FROM Salaries
WHERE Year = 2011'''

sql_query(q)
```

```python
### 5. Plot the top 10 job titles with the highest TotalPay in 2014.

q = '''SELECT JobTitle, SUM(TotalPay) AS TotalPay_Sum
FROM Salaries
WHERE Year = 2014
GROUP BY JobTitle
ORDER BY TotalPay_Sum DESC
LIMIT 10'''

sql_query(q)
```

```python
#### 6. Which job titles have overtime pay exceeding $150,000 exluding 'Not provided'

q = '''SELECT DISTINCT JobTitle, OvertimePay
FROM Salaries
WHERE OvertimePay > 150000 AND JobTitle NOT IN ('Not Provided', 'Not provided')
ORDER BY OvertimePay DESC'''

sql_query(q)
```

```python
### 7. Find all employees whose job titles contain the word “CHIEF”
q = '''
SELECT *
FROM Salaries
WHERE UPPER(JobTitle) LIKE '%CHIEF%'
'''
sql_query(q)
```

```python
### 8. Count how many employees have job titles containing “ENGINEER” and “Fire” in year 2012 and 2013

q = '''
SELECT Year, COUNT(*) AS EmployeeCount
FROM Salaries
WHERE 
    (UPPER(JobTitle) LIKE '%ENGINEER%' OR UPPER(JobTitle) LIKE '%FIRE%')
    AND Year IN (2012, 2013)
GROUP BY Year'''
sql_query(q)
```

```python
### 9. Distinct job titles starting with "CAPTAIN"

q = '''SELECT DISTINCT JobTitle
FROM Salaries
WHERE UPPER(JobTitle) LIKE 'CAPTAIN%'
'''
sql_query(q)
```

```python
### 10. Employees having basepay > 50k and totalpay < 150000 exlucding 'Not Provided'

q = '''
SELECT *
FROM Salaries
WHERE 
    BasePay NOT IN ('Not Provided', 'not provided') AND
    TotalPay NOT IN ('Not Provided', 'not provided') AND
    CAST(BasePay AS FLOAT) > 50000 AND
    CAST(TotalPay AS FLOAT) < 150000'''

sql_query(q)
```
