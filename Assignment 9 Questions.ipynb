### Question 3

- How can we detect and identify outliers in Daily Usage Hours, Exercise Hours, and Screen Time Before Bed among teens using the IQR method?
- Display the records corresponding to these outliers using boxplot.



```python

# Load the data (skip if already loaded)
df = pd.read_csv("teen_phone_addiction_dataset.csv")

# Columns to analyze
cols = ['Daily_Usage_Hours', 'Exercise_Hours', 'Screen_Time_Before_Bed']

# Drop rows with missing values for selected columns
df_iqr = df[cols].dropna()

# Create an empty DataFrame to collect outliers
outliers_df = pd.DataFrame()

# Loop through each column and find outliers using IQR
for col in cols:
    Q1 = df_iqr[col].quantile(0.25)
    Q3 = df_iqr[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Find outlier rows
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outliers_df = pd.concat([outliers_df, outliers])

    print(f"\n📊 Outliers in {col}:")
    print(outliers[[col]])

# Drop duplicates (in case rows are outliers in multiple columns)
outliers_df = outliers_df.drop_duplicates()

# Display the full outlier records
print("\n📋 All Outlier Records (any of the 3 columns):")
display(outliers_df)

# Plot boxplots to visualize outliers
plt.figure(figsize=(12, 6))
for i, col in enumerate(cols, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[col])
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
# Select only the numeric variables mentioned in the question
columns = [
    'Daily_Usage_Hours', 'Sleep_Hours', 'Academic_Performance', 'Addiction_Level',
    'Time_on_Social_Media', 'Time_on_Gaming', 'Time_on_Education', 'Apps_Used_Daily',
    'Anxiety_Level', 'Depression_Level', 'Self_Esteem', 'Social_Interactions',
    'Exercise_Hours', 'Screen_Time_Before_Bed', 'Phone_Checks_Per_Day',
    'Family_Communication', 'Weekend_Usage_Hours'
]

# Filter DataFrame
df_selected = df[columns]

# Compute correlation matrix
correlation_matrix = df_selected.corr()

# Show it
correlation_matrix

plt.figure(figsize=(14, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title("Correlation Heatmap of Teen Behavior and Screen Usage")
plt.show()
```

### Question 5

How does the average addiction level vary across different age groups? Show with line graph

```python
# Group by Age and calculate average addiction level
age_addiction = df.groupby('Age')['Addiction_Level'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.lineplot(data=age_addiction, x='Age', y='Addiction_Level', marker='o', linewidth=2.5, color='crimson')
plt.title('Average Addiction Level by Age Group', fontsize=14)
plt.xlabel('Age')
plt.ylabel('Average Addiction Level')
plt.grid(True)
plt.tight_layout()
plt.show()
```
